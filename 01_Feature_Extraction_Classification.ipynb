```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from math import pi

# Set global thesis style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.dpi'] = 300

def generate_f1_visuals(csv_path, task_name, best_model):
    try:
        df = pd.read_csv(csv_path)
    except FileNotFoundError:
        print(f"Could not find {csv_path}. Skipping.")
        return

    print(f"Generating visuals for {task_name}...")

    # --- IDEA 1: Champion's Ablation Chart ---
    plt.figure(figsize=(10, 6))
    champion_df = df[df['Model'] == best_model].copy()

    def make_label(row):
        features = []
        if row['ROM']: features.append('ROM')
        if row['Velocity']: features.append('Vel')
        if row['Symmetry']: features.append('Sym')
        if row['Feature Selection']: features.append('FS')
        return " + ".join(features) if features else "Baseline Only"

    champion_df['Combo_Label'] = champion_df.apply(make_label, axis=1)
    champion_df = champion_df.sort_values('F1-Measure', ascending=True).tail(8)

    colors = ['#d3d3d3'] * 7 + ['#2ca02c']
    bars = plt.barh(champion_df['Combo_Label'], champion_df['F1-Measure'], color=colors)

    for bar in bars:
        width = bar.get_width()
        plt.text(width - 0.01, bar.get_y() + bar.get_height()/2, f'{width:.4f}',
                 va='center', ha='right', color='white', fontweight='bold')

    plt.title(f"{best_model} Ablation Impact on F1-Measure\n({task_name} - 23 Joints)", fontweight='bold')
    plt.xlabel("F1-Measure")
    plt.xlim(max(0, min(champion_df['F1-Measure']) - 0.05), max(champion_df['F1-Measure']) + 0.02)
    plt.tight_layout()
    plt.savefig(f"{task_name.replace(' ', '_')}_Idea1_Champion_F1.png")
    plt.close()

    # --- IDEA 2: Universal Feature Power Histogram ---
    plt.figure(figsize=(8, 6))
    isolated_scores = {}
    features = ['ROM', 'Velocity', 'Symmetry', 'Feature Selection']

    for feat in features:
        other_feats = [f for f in features if f != feat]
        condition = (df[feat] == True) & (df[other_feats[0]] == False) & (df[other_feats[1]] == False) & (df[other_feats[2]] == False)
        isolated_scores[feat] = df[condition]['F1-Measure'].mean()

    baseline_mean = df[(df['ROM']==False) & (df['Velocity']==False) & (df['Symmetry']==False) & (df['Feature Selection']==False)]['F1-Measure'].mean()
    isolated_scores['Baseline Only'] = baseline_mean

    sorted_scores = dict(sorted(isolated_scores.items(), key=lambda item: item[1]))
    sns.barplot(x=list(sorted_scores.keys()), y=list(sorted_scores.values()), palette="viridis")
    plt.title(f"Universal Feature Power (Average Isolated F1)\n({task_name} - 23 Joints)", fontweight='bold')
    plt.ylabel("Average F1-Measure Across All 10 Models")
    plt.ylim(max(0, min(sorted_scores.values()) - 0.05), max(sorted_scores.values()) + 0.05)

    for i, v in enumerate(sorted_scores.values()):
        plt.text(i, v + 0.005, f"{v:.4f}", ha='center', fontweight='bold')

    plt.tight_layout()
    plt.savefig(f"{task_name.replace(' ', '_')}_Idea2_Histogram_F1.png")
    plt.close()

    # --- IDEA 3: The Search Space Heatmap ---
    plt.figure(figsize=(12, 8))
    df['Combo_Index'] = df.apply(lambda r: f"R:{int(r['ROM'])} V:{int(r['Velocity'])} S:{int(r['Symmetry'])} F:{int(r['Feature Selection'])}", axis=1)
    pivot_table = df.pivot(index="Model", columns="Combo_Index", values="F1-Measure")
    sns.heatmap(pivot_table, cmap="YlGnBu", cbar_kws={'label': 'F1-Measure'}, linewidths=.5)
    plt.title(f"Grid Search Heatmap (F1-Measure)\n{task_name} - 23 Joints", fontweight='bold')
    plt.xlabel("Feature Combinations (1=True, 0=False)")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(f"{task_name.replace(' ', '_')}_Idea3_Heatmap_F1.png")
    plt.close()

def get_top_feature_frequencies(csv_path, top_n=20):
    try:
        df = pd.read_csv(csv_path)
    except FileNotFoundError:
        return None
    top_df = df.sort_values('F1-Measure', ascending=False).head(top_n)
    features = ['ROM', 'Velocity', 'Symmetry', 'Feature Selection']
    return [(top_df[f] == True).sum() / top_n for f in features]

def plot_radar_chart():
    print("Generating Contrast Radar Chart...")
    task1_freqs = get_top_feature_frequencies('task1_full_matrix_leaderboard_F1.csv')
    task2_freqs = get_top_feature_frequencies('task2_full_matrix_leaderboard_F1.csv')

    if not task1_freqs or not task2_freqs:
        return

    categories = ['ROM', 'Velocity', 'Symmetry', 'Feature\nSelection']
    N = len(categories)
    angles = [n / float(N) * 2 * pi for n in range(N)]
    angles += angles[:1]
    task1_freqs += task1_freqs[:1]
    task2_freqs += task2_freqs[:1]

    sns.set_theme(style="white")
    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))
    plt.xticks(angles[:-1], categories, size=12, fontweight='bold')
    ax.set_rlabel_position(45)
    plt.yticks([0.25, 0.50, 0.75, 1.00], ["25%", "50%", "75%", "100%"], color="grey", size=10)
    plt.ylim(0, 1.05)

    ax.plot(angles, task1_freqs, linewidth=2, linestyle='solid', label='Task 1: Expertise', color='#1f77b4')
    ax.fill(angles, task1_freqs, '#1f77b4', alpha=0.1)
    ax.plot(angles, task2_freqs, linewidth=2, linestyle='solid', label='Task 2: Clinical Diagnosis', color='#d62728')
    ax.fill(angles, task2_freqs, '#d62728', alpha=0.25)

    plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1), prop={'size': 11, 'weight': 'bold'})
    plt.title("Feature Dependency Profile (23 Joints)\n(Presence in Top 20 Pipelines)", size=14, fontweight='bold', y=1.1)
    plt.tight_layout()
    plt.savefig("Idea4_Contrast_Radar_23_Joints.png", bbox_inches='tight')
    plt.close()

# 1. Generate Task 1 Visuals targeting SVM
generate_f1_visuals('task1_full_matrix_leaderboard_F1.csv', 'Task 1', 'Support Vector Machine')

# 2. Generate Task 2 Visuals targeting Random Forest
generate_f1_visuals('task2_full_matrix_leaderboard_F1.csv', 'Task 2', 'Random Forest')

# 3. Generate the Contrast Radar
plot_radar_chart()

print("All high-resolution 23-joint presentation visuals generated successfully!")
```

```python
%%writefile scripts/optimize_task2.py
import pandas as pd
import numpy as np
import glob
from pathlib import Path
import warnings
import itertools

# --- SKLEARN MODELS ---
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.naive_bayes import GaussianNB

from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import SelectFromModel

from imblearn.over_sampling import SMOTE
from imblearn.pipeline import make_pipeline as make_imb_pipeline

warnings.filterwarnings("ignore")

# --> UPDATED DIRECTORY FOR 23/24 JOINT DATASET <--
INPUT_CSV_DIR = Path("outputs_24_joints")

def create_summary_features(csv_path, use_rom, use_vel, use_sym):
    try:
        df = pd.read_csv(csv_path)
        if 'timestamp_sec' in df.columns:
            df = df.drop(columns=['timestamp_sec'])
        df = df.dropna()
        if df.empty or len(df) < 10: return None

        features = {}
        for col in df.columns:
            features[f'{col}_mean'] = df[col].mean()
            features[f'{col}_median'] = df[col].median()
            features[f'{col}_std'] = df[col].std()

            if use_rom:
                features[f'{col}_rom'] = df[col].max() - df[col].min()
            if use_vel:
                velocity = df[col].diff().abs()
                features[f'{col}_vel_mean'] = velocity.mean()
                features[f'{col}_vel_std'] = velocity.std()

        if use_sym:
            if 'L_Wrist_y' in df.columns and 'R_Wrist_y' in df.columns:
                l_rom = df['L_Wrist_y'].max() - df['L_Wrist_y'].min()
                r_rom = df['R_Wrist_y'].max() - df['R_Wrist_y'].min()
                features['Wrist_Y_Symmetry'] = abs(l_rom - r_rom)
            if 'L_Knee_y' in df.columns and 'R_Knee_y' in df.columns:
                l_k_rom = df['L_Knee_y'].max() - df['L_Knee_y'].min()
                r_k_rom = df['R_Knee_y'].max() - df['R_Knee_y'].min()
                features['Knee_Y_Symmetry'] = abs(l_k_rom - r_k_rom)
        return features
    except Exception:
        return None

def main():
    csv_files = glob.glob(str(INPUT_CSV_DIR / "*.csv"))
    if not csv_files:
        print(f"Error: No CSV files found in {INPUT_CSV_DIR}")
        return

    options = [True, False]
    combinations = list(itertools.product(options, options, options, options))

    cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
    scoring_metrics = ['accuracy', 'f1_macro', 'precision_macro', 'recall_macro']

    models = {
        "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
        "Extra Trees": ExtraTreesClassifier(n_estimators=100, random_state=42),
        "AdaBoost": AdaBoostClassifier(random_state=42),
        "Decision Tree": DecisionTreeClassifier(random_state=42),
        "Gradient Boosting": GradientBoostingClassifier(random_state=42),
        "KNN": KNeighborsClassifier(),
        "Linear Discriminant Analysis": LinearDiscriminantAnalysis(),
        "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
        "Naive Bayes": GaussianNB(),
        "Support Vector Machine": SVC(random_state=42)
    }

    print("=" * 110)
    print("STARTING F1-OPTIMIZED GRID SEARCH: TASK 2 (Stroke, BackPain, Parkinson) - 23 JOINTS")
    print(f"Testing 16 feature combinations across 10 models (1,600 training cycles)...")
    print("=" * 110)

    leaderboard = []

    for idx, (use_rom, use_vel, use_sym, use_fs) in enumerate(combinations):
        X, y = [], []
        for csv_file in csv_files:
            basename = Path(csv_file).name
            group_label = basename.split('_')[0]

            if group_label in ['S', 'B', 'P']:
                summary = create_summary_features(csv_file, use_rom, use_vel, use_sym)
                if summary:
                    X.append(summary)
                    y.append(group_label)

        if not X: continue

        df_X = pd.DataFrame(X)
        y_encoded = LabelEncoder().fit_transform(y)

        for model_name, model in models.items():
            steps = [SMOTE(random_state=42), StandardScaler()]
            if use_fs:
                steps.append(SelectFromModel(RandomForestClassifier(n_estimators=50, random_state=42)))
            steps.append(model)

            pipeline = make_imb_pipeline(*steps)
            scores = cross_validate(pipeline, df_X, y_encoded, cv=cv, scoring=scoring_metrics, n_jobs=-1)

            leaderboard.append({
                'F1-Measure': scores['test_f1_macro'].mean(),
                'Accuracy': scores['test_accuracy'].mean(),
                'Precision': scores['test_precision_macro'].mean(),
                'Recall': scores['test_recall_macro'].mean(),
                'Model': model_name,
                'ROM': use_rom,
                'Velocity': use_vel,
                'Symmetry': use_sym,
                'Feature Selection': use_fs
            })

        print(f"Completed Combination {idx+1}/16")

    df_leaderboard = pd.DataFrame(leaderboard).sort_values(by="F1-Measure", ascending=False).reset_index(drop=True)

    display_df = df_leaderboard.copy()
    display_df['F1-Measure'] = display_df['F1-Measure'].map('{:.4f}'.format)
    display_df['Accuracy'] = display_df['Accuracy'].map('{:.4f}'.format)
    display_df['Precision'] = display_df['Precision'].map('{:.4f}'.format)
    display_df['Recall'] = display_df['Recall'].map('{:.4f}'.format)

    print("\n" + "="*110)
    print("🏆 FINAL F1-RANKED LEADERBOARD: TASK 2 (23 JOINTS) 🏆")
    print("="*110)
    print(display_df.head(15).to_string(index=False))

    filename = "task2_full_matrix_leaderboard_F1.csv"
    df_leaderboard.to_csv(filename, index=False)
    print(f"\nFull results saved to {filename} (Overwritten with 23-joint data)")

if __name__ == "__main__":
    main()
```

```python
%%writefile scripts/optimize_task1.py
import pandas as pd
import numpy as np
import glob
from pathlib import Path
import warnings
import itertools

# --- SKLEARN MODELS ---
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.naive_bayes import GaussianNB

from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import SelectFromModel

from imblearn.over_sampling import SMOTE
from imblearn.pipeline import make_pipeline as make_imb_pipeline

warnings.filterwarnings("ignore")

# --> UPDATED DIRECTORY FOR 23/24 JOINT DATASET <--
INPUT_CSV_DIR = Path("outputs_24_joints")

def create_summary_features(csv_path, use_rom, use_vel, use_sym):
    try:
        df = pd.read_csv(csv_path)
        if 'timestamp_sec' in df.columns:
            df = df.drop(columns=['timestamp_sec'])
        df = df.dropna()
        if df.empty or len(df) < 10: return None

        features = {}
        for col in df.columns:
            features[f'{col}_mean'] = df[col].mean()
            features[f'{col}_median'] = df[col].median()
            features[f'{col}_std'] = df[col].std()

            if use_rom:
                features[f'{col}_rom'] = df[col].max() - df[col].min()
            if use_vel:
                velocity = df[col].diff().abs()
                features[f'{col}_vel_mean'] = velocity.mean()
                features[f'{col}_vel_std'] = velocity.std()

        if use_sym:
            # Safely checks if wrists/knees are still in the 23-joint dataset
            if 'L_Wrist_y' in df.columns and 'R_Wrist_y' in df.columns:
                l_rom = df['L_Wrist_y'].max() - df['L_Wrist_y'].min()
                r_rom = df['R_Wrist_y'].max() - df['R_Wrist_y'].min()
                features['Wrist_Y_Symmetry'] = abs(l_rom - r_rom)
            if 'L_Knee_y' in df.columns and 'R_Knee_y' in df.columns:
                l_k_rom = df['L_Knee_y'].max() - df['L_Knee_y'].min()
                r_k_rom = df['R_Knee_y'].max() - df['R_Knee_y'].min()
                features['Knee_Y_Symmetry'] = abs(l_k_rom - r_k_rom)
        return features
    except Exception:
        return None

def main():
    csv_files = glob.glob(str(INPUT_CSV_DIR / "*.csv"))
    if not csv_files:
        print(f"Error: No CSV files found in {INPUT_CSV_DIR}")
        return

    options = [True, False]
    combinations = list(itertools.product(options, options, options, options))

    cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
    scoring_metrics = ['accuracy', 'f1_macro', 'precision_macro', 'recall_macro']

    models = {
        "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
        "Extra Trees": ExtraTreesClassifier(n_estimators=100, random_state=42),
        "AdaBoost": AdaBoostClassifier(random_state=42),
        "Decision Tree": DecisionTreeClassifier(random_state=42),
        "Gradient Boosting": GradientBoostingClassifier(random_state=42),
        "KNN": KNeighborsClassifier(),
        "Linear Discriminant Analysis": LinearDiscriminantAnalysis(),
        "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
        "Naive Bayes": GaussianNB(),
        "Support Vector Machine": SVC(random_state=42)
    }

    print("=" * 100)
    print("STARTING F1-OPTIMIZED GRID SEARCH: TASK 1 (Expert vs Non-Expert) - 23 JOINTS")
    print(f"Testing 16 feature combinations across 10 models (1,600 training cycles)...")
    print("=" * 100)

    leaderboard = []

    for idx, (use_rom, use_vel, use_sym, use_fs) in enumerate(combinations):
        X, y = [], []
        for csv_file in csv_files:
            basename = Path(csv_file).name
            group_label = basename.split('_')[0]
            if group_label in ['E', 'NE']:
                summary = create_summary_features(csv_file, use_rom, use_vel, use_sym)
                if summary:
                    X.append(summary)
                    y.append(group_label)

        if not X: continue

        df_X = pd.DataFrame(X)
        y_encoded = LabelEncoder().fit_transform(y)

        for model_name, model in models.items():
            steps = [SMOTE(random_state=42), StandardScaler()]
            if use_fs:
                steps.append(SelectFromModel(RandomForestClassifier(n_estimators=50, random_state=42)))
            steps.append(model)

            pipeline = make_imb_pipeline(*steps)
            scores = cross_validate(pipeline, df_X, y_encoded, cv=cv, scoring=scoring_metrics, n_jobs=-1)

            leaderboard.append({
                'F1-Measure': scores['test_f1_macro'].mean(),
                'Accuracy': scores['test_accuracy'].mean(),
                'Precision': scores['test_precision_macro'].mean(),
                'Recall': scores['test_recall_macro'].mean(),
                'Model': model_name,
                'ROM': use_rom,
                'Velocity': use_vel,
                'Symmetry': use_sym,
                'Feature Selection': use_fs
            })

        print(f"Completed Combination {idx+1}/16")

    df_leaderboard = pd.DataFrame(leaderboard).sort_values(by="F1-Measure", ascending=False).reset_index(drop=True)

    display_df = df_leaderboard.copy()
    display_df['F1-Measure'] = display_df['F1-Measure'].map('{:.4f}'.format)
    display_df['Accuracy'] = display_df['Accuracy'].map('{:.4f}'.format)
    display_df['Precision'] = display_df['Precision'].map('{:.4f}'.format)
    display_df['Recall'] = display_df['Recall'].map('{:.4f}'.format)

    print("\n" + "="*110)
    print("🏆 FINAL F1-RANKED LEADERBOARD: TASK 1 (23 JOINTS) 🏆")
    print("="*110)
    print(display_df.head(15).to_string(index=False))

    filename = "task1_full_matrix_leaderboard_F1.csv"
    df_leaderboard.to_csv(filename, index=False)
    print(f"\nFull results saved to {filename} (Overwritten with 23-joint data)")

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from math import pi

# Set global thesis style
sns.set_theme(style="white")
plt.rcParams['figure.dpi'] = 300

def get_top_feature_frequencies(csv_path, top_n=20):
    """Calculates the percentage of top pipelines that use each feature."""
    try:
        df = pd.read_csv(csv_path)
    except FileNotFoundError:
        print(f"Could not find {csv_path}.")
        return None

    # Get the top N pipelines
    top_df = df.sort_values('F1-Measure', ascending=False).head(top_n)

    features = ['ROM', 'Velocity', 'Symmetry', 'Feature Selection']
    frequencies = []

    for f in features:
        # Calculate percentage (0.0 to 1.0)
        freq = (top_df[f] == True).sum() / top_n
        frequencies.append(freq)

    return frequencies

def plot_radar_chart():
    # 1. Get the data
    task1_freqs = get_top_feature_frequencies('task1_full_matrix_leaderboard_F1.csv')
    task2_freqs = get_top_feature_frequencies('task2_full_matrix_leaderboard_F1.csv')

    if not task1_freqs or not task2_freqs:
        return

    # 2. Set up the radar chart parameters
    categories = ['ROM', 'Velocity', 'Symmetry', 'Feature\nSelection']
    N = len(categories)

    # Calculate angles for each category
    angles = [n / float(N) * 2 * pi for n in range(N)]
    angles += angles[:1] # Close the loop

    # Close the loop for the data arrays
    task1_freqs += task1_freqs[:1]
    task2_freqs += task2_freqs[:1]

    # 3. Create the plot
    fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

    # Draw one axe per variable and add labels
    plt.xticks(angles[:-1], categories, size=12, fontweight='bold')

    # Draw ylabels
    ax.set_rlabel_position(45)
    plt.yticks([0.25, 0.50, 0.75, 1.00], ["25%", "50%", "75%", "100%"], color="grey", size=10)
    plt.ylim(0, 1.05)

    # 4. Plot Task 1 (Expertise)
    ax.plot(angles, task1_freqs, linewidth=2, linestyle='solid', label='Task 1: Expertise', color='#1f77b4')
    ax.fill(angles, task1_freqs, '#1f77b4', alpha=0.1)

    # 5. Plot Task 2 (Clinical Diagnosis)
    ax.plot(angles, task2_freqs, linewidth=2, linestyle='solid', label='Task 2: Clinical Diagnosis', color='#d62728')
    ax.fill(angles, task2_freqs, '#d62728', alpha=0.25)

    # Add legend and title
    plt.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1), prop={'size': 11, 'weight': 'bold'})
    plt.title("Feature Dependency Profile\n(Presence in Top 20 Pipelines)", size=14, fontweight='bold', y=1.1)

    plt.tight_layout()
    plt.savefig("Idea4_Contrast_Radar.png", bbox_inches='tight')
    print("Successfully generated Idea4_Contrast_Radar.png!")
    plt.close()

# Execute the chart generator
plot_radar_chart()
```

```python
### Idea1-2-3 figures
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Set global thesis style for professional presentation
sns.set_theme(style="whitegrid")
plt.rcParams['figure.dpi'] = 300

def generate_f1_visuals(csv_path, task_name, best_model):
    try:
        df = pd.read_csv(csv_path)
    except FileNotFoundError:
        print(f"Could not find {csv_path}. Make sure you ran the F1 scripts.")
        return

    print(f"\nGenerating F1-based visuals for {task_name}...")

    # ---------------------------------------------------------
    # IDEA 1: Champion's Ablation Chart (F1-Measure)
    # ---------------------------------------------------------
    plt.figure(figsize=(10, 6))
    champion_df = df[df['Model'] == best_model].copy()

    def make_label(row):
        features = []
        if row['ROM']: features.append('ROM')
        if row['Velocity']: features.append('Vel')
        if row['Symmetry']: features.append('Sym')
        if row['Feature Selection']: features.append('FS')
        return " + ".join(features) if features else "Baseline Only"

    champion_df['Combo_Label'] = champion_df.apply(make_label, axis=1)
    # Get the top 8 combinations for this specific model to show the drop-off
    champion_df = champion_df.sort_values('F1-Measure', ascending=True).tail(8)

    colors = ['#d3d3d3'] * 7 + ['#2ca02c'] # Highlight the ultimate top combo in green
    bars = plt.barh(champion_df['Combo_Label'], champion_df['F1-Measure'], color=colors)

    for bar in bars:
        width = bar.get_width()
        plt.text(width - 0.01, bar.get_y() + bar.get_height()/2, f'{width:.4f}',
                 va='center', ha='right', color='white', fontweight='bold')

    plt.title(f"{best_model} Feature Ablation Impact on F1-Measure\n({task_name})", fontweight='bold')
    plt.xlabel("F1-Measure")
    plt.xlim(max(0, min(champion_df['F1-Measure']) - 0.05), max(champion_df['F1-Measure']) + 0.02)
    plt.tight_layout()
    plt.savefig(f"{task_name.replace(' ', '_')}_Idea1_Champion_F1.png")
    plt.close()

    # ---------------------------------------------------------
    # IDEA 2: Universal Feature Power Histogram (F1-Measure)
    # ---------------------------------------------------------
    plt.figure(figsize=(8, 6))

    isolated_scores = {}
    features = ['ROM', 'Velocity', 'Symmetry', 'Feature Selection']

    for feat in features:
        other_feats = [f for f in features if f != feat]
        condition = (df[feat] == True) & (df[other_feats[0]] == False) & (df[other_feats[1]] == False) & (df[other_feats[2]] == False)
        isolated_mean = df[condition]['F1-Measure'].mean()
        isolated_scores[feat] = isolated_mean

    baseline_mean = df[(df['ROM']==False) & (df['Velocity']==False) & (df['Symmetry']==False) & (df['Feature Selection']==False)]['F1-Measure'].mean()
    isolated_scores['Baseline Only'] = baseline_mean

    sorted_scores = dict(sorted(isolated_scores.items(), key=lambda item: item[1]))

    sns.barplot(x=list(sorted_scores.keys()), y=list(sorted_scores.values()), palette="viridis")
    plt.title(f"Universal Feature Power (Average Isolated F1-Measure)\n({task_name})", fontweight='bold')
    plt.ylabel("Average F1-Measure Across All 10 Models")
    plt.ylim(max(0, min(sorted_scores.values()) - 0.05), max(sorted_scores.values()) + 0.05)

    for i, v in enumerate(sorted_scores.values()):
        plt.text(i, v + 0.005, f"{v:.4f}", ha='center', fontweight='bold')

    plt.tight_layout()
    plt.savefig(f"{task_name.replace(' ', '_')}_Idea2_Histogram_F1.png")
    plt.close()

    # ---------------------------------------------------------
    # IDEA 3: The Search Space Heatmap (F1-Measure)
    # ---------------------------------------------------------
    plt.figure(figsize=(12, 8))

    df['Combo_Index'] = df.apply(lambda r: f"R:{int(r['ROM'])} V:{int(r['Velocity'])} S:{int(r['Symmetry'])} F:{int(r['Feature Selection'])}", axis=1)

    pivot_table = df.pivot(index="Model", columns="Combo_Index", values="F1-Measure")
    sns.heatmap(pivot_table, cmap="YlGnBu", cbar_kws={'label': 'F1-Measure'}, linewidths=.5)

    plt.title(f"Full-Matrix Grid Search Heatmap (F1-Measure)\n{task_name}", fontweight='bold')
    plt.xlabel("Feature Combinations (1=True, 0=False)")
    plt.xticks(rotation=45, ha='right')
    plt.tight_layout()
    plt.savefig(f"{task_name.replace(' ', '_')}_Idea3_Heatmap_F1.png")
    plt.close()

    print(f"Successfully saved 3 high-res charts for {task_name}.")

# 1. Generate Task 1 Visuals targeting SVM
generate_f1_visuals('task1_full_matrix_leaderboard_F1.csv', 'Task 1', 'Support Vector Machine')

# 2. Generate Task 2 Visuals targeting Random Forest
generate_f1_visuals('task2_full_matrix_leaderboard_F1.csv', 'Task 2', 'Random Forest')
```

```python
### Task 1 için tüm configuration kombinasyonları testi (tüm metrikler)
%%writefile scripts/optimize_task1.py
import pandas as pd
import numpy as np
import glob
from pathlib import Path
import warnings
import itertools

# --- SKLEARN MODELS ---
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.naive_bayes import GaussianNB

from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import SelectFromModel

from imblearn.over_sampling import SMOTE
from imblearn.pipeline import make_pipeline as make_imb_pipeline

warnings.filterwarnings("ignore")
INPUT_CSV_DIR = Path("outputs")

def create_summary_features(csv_path, use_rom, use_vel, use_sym):
    try:
        df = pd.read_csv(csv_path)
        if 'timestamp_sec' in df.columns:
            df = df.drop(columns=['timestamp_sec'])
        df = df.dropna()
        if df.empty or len(df) < 10: return None

        features = {}
        for col in df.columns:
            features[f'{col}_mean'] = df[col].mean()
            features[f'{col}_median'] = df[col].median()
            features[f'{col}_std'] = df[col].std()

            if use_rom:
                features[f'{col}_rom'] = df[col].max() - df[col].min()
            if use_vel:
                velocity = df[col].diff().abs()
                features[f'{col}_vel_mean'] = velocity.mean()
                features[f'{col}_vel_std'] = velocity.std()

        if use_sym:
            if 'L_Wrist_y' in df.columns and 'R_Wrist_y' in df.columns:
                l_rom = df['L_Wrist_y'].max() - df['L_Wrist_y'].min()
                r_rom = df['R_Wrist_y'].max() - df['R_Wrist_y'].min()
                features['Wrist_Y_Symmetry'] = abs(l_rom - r_rom)
            if 'L_Knee_y' in df.columns and 'R_Knee_y' in df.columns:
                l_k_rom = df['L_Knee_y'].max() - df['L_Knee_y'].min()
                r_k_rom = df['R_Knee_y'].max() - df['R_Knee_y'].min()
                features['Knee_Y_Symmetry'] = abs(l_k_rom - r_k_rom)
        return features
    except Exception:
        return None

def main():
    csv_files = glob.glob(str(INPUT_CSV_DIR / "*.csv"))
    if not csv_files: return

    options = [True, False]
    combinations = list(itertools.product(options, options, options, options))

    cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
    # Added F1, Precision, and Recall to the tracking metrics
    scoring_metrics = ['accuracy', 'f1_macro', 'precision_macro', 'recall_macro']

    models = {
        "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
        "Extra Trees": ExtraTreesClassifier(n_estimators=100, random_state=42),
        "AdaBoost": AdaBoostClassifier(random_state=42),
        "Decision Tree": DecisionTreeClassifier(random_state=42),
        "Gradient Boosting": GradientBoostingClassifier(random_state=42),
        "KNN": KNeighborsClassifier(),
        "Linear Discriminant Analysis": LinearDiscriminantAnalysis(),
        "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
        "Naive Bayes": GaussianNB(),
        "Support Vector Machine": SVC(random_state=42)
    }

    print("=" * 100)
    print("STARTING F1-OPTIMIZED GRID SEARCH: TASK 1 (Expert vs Non-Expert)")
    print(f"Testing 16 feature combinations across 10 models (1,600 training cycles)...")
    print("=" * 100)

    leaderboard = []

    for idx, (use_rom, use_vel, use_sym, use_fs) in enumerate(combinations):
        X, y = [], []
        for csv_file in csv_files:
            basename = Path(csv_file).name
            group_label = basename.split('_')[0]
            if group_label in ['E', 'NE']:
                summary = create_summary_features(csv_file, use_rom, use_vel, use_sym)
                if summary:
                    X.append(summary)
                    y.append(group_label)

        if not X: continue

        df_X = pd.DataFrame(X)
        y_encoded = LabelEncoder().fit_transform(y)

        for model_name, model in models.items():
            steps = [SMOTE(random_state=42), StandardScaler()]
            if use_fs:
                steps.append(SelectFromModel(RandomForestClassifier(n_estimators=50, random_state=42)))
            steps.append(model)

            pipeline = make_imb_pipeline(*steps)
            scores = cross_validate(pipeline, df_X, y_encoded, cv=cv, scoring=scoring_metrics, n_jobs=-1)

            # Extracting all four metrics
            accuracy = scores['test_accuracy'].mean()
            f1 = scores['test_f1_macro'].mean()
            precision = scores['test_precision_macro'].mean()
            recall = scores['test_recall_macro'].mean()

            leaderboard.append({
                'F1-Measure': f1,
                'Accuracy': accuracy,
                'Precision': precision,
                'Recall': recall,
                'Model': model_name,
                'ROM': use_rom,
                'Velocity': use_vel,
                'Symmetry': use_sym,
                'Feature Selection': use_fs
            })

        print(f"Completed Combination {idx+1}/16")

    # Sorting strictly by F1-Measure
    df_leaderboard = pd.DataFrame(leaderboard).sort_values(by="F1-Measure", ascending=False).reset_index(drop=True)

    # Format the numbers for clean printing
    display_df = df_leaderboard.copy()
    display_df['F1-Measure'] = display_df['F1-Measure'].map('{:.4f}'.format)
    display_df['Accuracy'] = display_df['Accuracy'].map('{:.4f}'.format)
    display_df['Precision'] = display_df['Precision'].map('{:.4f}'.format)
    display_df['Recall'] = display_df['Recall'].map('{:.4f}'.format)

    print("\n" + "="*110)
    print("🏆 FINAL F1-RANKED LEADERBOARD: TASK 1 (TOP 15) 🏆")
    print("="*110)
    print(display_df.head(15).to_string(index=False))

    filename = "task1_full_matrix_leaderboard_F1.csv"
    df_leaderboard.to_csv(filename, index=False)
    print(f"\nFull results saved to {filename}")

if __name__ == "__main__":
    main()
```

```python
%%writefile scripts/feature_extractor_24.py
import cv2
import argparse
import sys
from pathlib import Path
from mediapipe.tasks.python.vision import pose_landmarker
from mediapipe.tasks.python.vision.core import image as mp_image_lib
from mediapipe.tasks.python.core import base_options
from mediapipe.tasks.python.vision.core import vision_task_running_mode

# --- THE 24-JOINT UPGRADE ---
# We keep the core 12, and add head posture (0, 2, 5),
# distal arms (17, 18, 19, 20), and distal legs (29, 30, 31, 32).
TRACKED_LANDMARKS = {
    0: "Nose",         2: "R_Eye",        5: "L_Eye",
    11: "L_Shoulder", 12: "R_Shoulder",
    13: "L_Elbow",    14: "R_Elbow",
    15: "L_Wrist",    16: "R_Wrist",
    17: "L_Pinky",    18: "R_Pinky",
    19: "L_Index",    20: "R_Index",
    23: "L_Hip",      24: "R_Hip",
    25: "L_Knee",     26: "R_Knee",
    27: "L_Ankle",    28: "R_Ankle",
    29: "L_Heel",     30: "R_Heel",
    31: "L_Foot_Idx", 32: "R_Foot_Idx"
}

OUTPUTS_DIR = Path("outputs_24_joints") # NEW SECURE FOLDER

SCRIPT_DIR = Path(__file__).resolve().parent
DEFAULT_POSE_MODEL_PATH = SCRIPT_DIR / "pose_landmarker_heavy.task"

def _ensure_pose_model() -> Path:
    path = DEFAULT_POSE_MODEL_PATH
    if path.is_file(): return path
    print("Heavy Pose modeli indiriliyor...", flush=True)
    import urllib.request
    urllib.request.urlretrieve("https://storage.googleapis.com/mediapipe-models/pose_landmarker/pose_landmarker_heavy/float16/latest/pose_landmarker_heavy.task", path)
    return path

def video_path_to_csv_basename(video_path: Path, data_path: Path) -> str:
    try: rel = video_path.resolve().relative_to(data_path.resolve())
    except ValueError: rel = video_path
    if len(rel.parts) <= 1: return rel.stem + ".csv"
    return "_".join(rel.parts[:-2] + (rel.stem,)) + ".csv"

def process_one_video(video_path, output_csv_path=None, is_batch=False):
    cap = cv2.VideoCapture(str(video_path))
    if not cap.isOpened(): return 0

    model_bytes = open(_ensure_pose_model(), "rb").read()
    opts = pose_landmarker.PoseLandmarkerOptions(
        base_options=base_options.BaseOptions(model_asset_buffer=model_bytes),
        running_mode=vision_task_running_mode.VisionTaskRunningMode.VIDEO,
        min_pose_detection_confidence=0.7,
        min_tracking_confidence=0.7,
    )
    landmarker = pose_landmarker.PoseLandmarker.create_from_options(opts)

    csv_file = None
    if output_csv_path is not None:
        csv_file = open(output_csv_path, "w", encoding="utf-8")
        header_cols = ["timestamp_sec"]
        for lm_id, name in TRACKED_LANDMARKS.items():
            header_cols.extend([f"{name}_x", f"{name}_y", f"{name}_z"])
        csv_file.write(",".join(header_cols) + "\n")

    frame_idx = 0
    fps = cap.get(cv2.CAP_PROP_FPS)
    if not fps or fps <= 0: fps = 30.0

    try:
        while True:
            ret, frame = cap.read()
            if not ret: break
            frame_idx += 1
            if is_batch and frame_idx % 50 == 0:
                print(f"    -> {frame_idx}. kare isleniyor...", end="\r", flush=True)

            timestamp_ms = int((frame_idx * 1000) / fps)
            rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            mp_img = mp_image_lib.Image(mp_image_lib.ImageFormat.SRGB, rgb)
            result = landmarker.detect_for_video(mp_img, timestamp_ms)

            row_values = [f"{timestamp_ms / 1000.0:.3f}"]

            if result.pose_world_landmarks and len(result.pose_world_landmarks) > 0:
                world = result.pose_world_landmarks[0]
                for lm_id in TRACKED_LANDMARKS.keys():
                    row_values.extend([f"{world[lm_id].x:.6f}", f"{world[lm_id].y:.6f}", f"{world[lm_id].z:.6f}"])
            else:
                row_values.extend([""] * (len(TRACKED_LANDMARKS) * 3))

            if csv_file: csv_file.write(",".join(row_values) + "\n")

        if is_batch: print(" " * 50, end="\r", flush=True)

    finally:
        cap.release()
        if csv_file: csv_file.close()
        landmarker.close()
    return frame_idx

def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("video", type=str, nargs="?", default=None)
    parser.add_argument("--data-path", type=str, default=None)
    args = parser.parse_args()

    if args.data_path is not None:
        data_path = Path(args.data_path)
        # ONLY GRAB MP4 FILES to save memory and time
        video_files = sorted(data_path.rglob("*.mp4"))

        OUTPUTS_DIR.mkdir(parents=True, exist_ok=True)
        print(f"\nToplam {len(video_files)} adet video bulundu. 24-EKLEM CIKARIMI basliyor...\n")

        for i, v in enumerate(video_files, 1):
            out_csv = OUTPUTS_DIR / video_path_to_csv_basename(v, data_path)

            if out_csv.exists() and out_csv.stat().st_size > 0:
                print(f"[{i}/{len(video_files)}] ATLANDI (Zaten hazir): {out_csv.stem}")
                continue

            print(f"[{i}/{len(video_files)}] Isleniyor: {out_csv.stem}")
            n = process_one_video(v, out_csv, is_batch=True)
            print(f"  => TAMAMLANDI! ({n} kare)\n")
        print("Tebrikler! Yeni 24-eklem CSV veri seti hazirlandi.")

if __name__ == "__main__":
    main()
```

```python
### Feature configuration teknikleri task specific hale getirildi
%%writefile scripts/train_model.py
import pandas as pd
import numpy as np
import glob
from pathlib import Path
import warnings

# --- SKLEARN MODELS ---
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier, AdaBoostClassifier, GradientBoostingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression, RidgeClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.naive_bayes import GaussianNB

from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.feature_selection import SelectFromModel

from imblearn.over_sampling import SMOTE
from imblearn.pipeline import make_pipeline as make_imb_pipeline

warnings.filterwarnings("ignore")
INPUT_CSV_DIR = Path("outputs")

# --- TASK-SPECIFIC CONFIGURATIONS ---
TASK_1_CONFIG = {
    'USE_BASELINE': True,
    'USE_ROM': True,
    'USE_VELOCITY': True,    # Task 1 needs full-body coordination (Velocity)
    'USE_SYMMETRY': False,
    'USE_FEATURE_SELECTION': False # No filtering, it needs all joints
}

TASK_2_CONFIG = {
    'USE_BASELINE': True,
    'USE_ROM': True,         # Task 2 needs to see physical limits (ROM)
    'USE_VELOCITY': False,   # Velocity creates camera noise here
    'USE_SYMMETRY': False,
    'USE_FEATURE_SELECTION': True  # Filters out irrelevant joints (Nose, ears, etc.)
}

def create_summary_features(csv_path, config):
    """Generates features based on the specific task configuration passed to it."""
    try:
        df = pd.read_csv(csv_path)
        if 'timestamp_sec' in df.columns:
            df = df.drop(columns=['timestamp_sec'])

        df = df.dropna()
        if df.empty or len(df) < 10: return None

        features = {}
        for col in df.columns:
            if config['USE_BASELINE']:
                features[f'{col}_mean'] = df[col].mean()
                features[f'{col}_median'] = df[col].median()
                features[f'{col}_std'] = df[col].std()
            if config['USE_ROM']:
                features[f'{col}_rom'] = df[col].max() - df[col].min()
            if config['USE_VELOCITY']:
                velocity = df[col].diff().abs()
                features[f'{col}_vel_mean'] = velocity.mean()
                features[f'{col}_vel_std'] = velocity.std()

        if config['USE_SYMMETRY']:
            if 'L_Wrist_y' in df.columns and 'R_Wrist_y' in df.columns:
                l_rom = df['L_Wrist_y'].max() - df['L_Wrist_y'].min()
                r_rom = df['R_Wrist_y'].max() - df['R_Wrist_y'].min()
                features['Wrist_Y_Symmetry'] = abs(l_rom - r_rom)
            if 'L_Knee_y' in df.columns and 'R_Knee_y' in df.columns:
                l_k_rom = df['L_Knee_y'].max() - df['L_Knee_y'].min()
                r_k_rom = df['R_Knee_y'].max() - df['R_Knee_y'].min()
                features['Knee_Y_Symmetry'] = abs(l_k_rom - r_k_rom)

        return features
    except Exception as e:
        return None

def evaluate_models(X, y, task_name, config):
    if len(X) == 0: return

    print(f"\nEvaluating {task_name}...")
    df_X = pd.DataFrame(X)
    y_encoded = LabelEncoder().fit_transform(y)

    cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
    scoring_metrics = ['accuracy', 'f1_macro', 'precision_macro', 'recall_macro']

    models = {
        "Random Forest": RandomForestClassifier(n_estimators=100, random_state=42),
        "Extra Trees": ExtraTreesClassifier(n_estimators=100, random_state=42),
        "AdaBoost": AdaBoostClassifier(random_state=42),
        "Decision Tree": DecisionTreeClassifier(random_state=42),
        "Gradient Boosting": GradientBoostingClassifier(random_state=42),
        "KNN": KNeighborsClassifier(),
        "Linear Discriminant Analysis": LinearDiscriminantAnalysis(),
        "Logistic Regression": LogisticRegression(max_iter=1000, random_state=42),
        "Naive Bayes": GaussianNB(),
        "Support Vector Machine": SVC(random_state=42)
    }

    results = []

    for name, model in models.items():
        steps = [SMOTE(random_state=42), StandardScaler()]

        # Apply Gini Decrease ONLY if the specific task config calls for it
        if config['USE_FEATURE_SELECTION']:
            steps.append(SelectFromModel(RandomForestClassifier(n_estimators=50, random_state=42)))

        steps.append(model)
        pipeline = make_imb_pipeline(*steps)
        scores = cross_validate(pipeline, df_X, y_encoded, cv=cv, scoring=scoring_metrics, n_jobs=-1)

        results.append({
            "Classifier": name,
            "Accuracy": f"{scores['test_accuracy'].mean():.4f}",
            "F-Measure": f"{scores['test_f1_macro'].mean():.4f}",
            "Precision": f"{scores['test_precision_macro'].mean():.4f}",
            "Recall": f"{scores['test_recall_macro'].mean():.4f}"
        })

    results_df = pd.DataFrame(results).sort_values(by="Accuracy", ascending=False).reset_index(drop=True)
    print("\n" + "="*80)
    print(f"COMPARATIVE ANALYSIS TABLE: {task_name}")
    print("="*80)
    print(results_df.to_string(index=False))

    filename = f"comparative_table_{task_name.replace(' ', '_')}.csv"
    results_df.to_csv(filename, index=False)
    print(f"\nSaved table to {filename}")

def main():
    csv_files = glob.glob(str(INPUT_CSV_DIR / "*.csv"))
    if not csv_files: return

    X_task1, y_task1 = [], []
    X_task2, y_task2 = [], []

    for csv_file in csv_files:
        basename = Path(csv_file).name
        group_label = basename.split('_')[0]

        # Route the file to the correct config based on the patient group
        if group_label in ['E', 'NE']:
            summary = create_summary_features(csv_file, TASK_1_CONFIG)
            if summary:
                X_task1.append(summary)
                y_task1.append(group_label)
        elif group_label in ['S', 'B', 'P']:
            summary = create_summary_features(csv_file, TASK_2_CONFIG)
            if summary:
                X_task2.append(summary)
                y_task2.append(group_label)

    evaluate_models(X_task1, y_task1, "Task 1 (Expert vs Non-Expert)", TASK_1_CONFIG)
    evaluate_models(X_task2, y_task2, "Task 2 (Stroke, BackPain, Parkinson)", TASK_2_CONFIG)

if __name__ == "__main__":
    main()
```

```python
import pandas as pd
import numpy as np
import glob
from pathlib import Path
import warnings

from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.pipeline import make_pipeline

warnings.filterwarnings("ignore")
INPUT_CSV_DIR = Path("outputs")

# --- ABLATION STUDY CONTROLS ---
# Turn these to True one by one to see their impact!
USE_BASELINE = True
USE_ROM = True
USE_VELOCITY = False
USE_SYMMETRY = True

def create_summary_features(csv_path):
    try:
        df = pd.read_csv(csv_path)
        if 'timestamp_sec' in df.columns:
            df = df.drop(columns=['timestamp_sec'])

        df = df.dropna()
        if df.empty or len(df) < 10:
            return None

        features = {}

        for col in df.columns:
            # 0. BASELINE (Raw spatial data - prone to camera distance bias)
            if USE_BASELINE:
                features[f'{col}_mean'] = df[col].mean()
                features[f'{col}_std'] = df[col].std()
                features[f'{col}_median'] = df[col].median()

            # 1. RANGE OF MOTION (Total travel distance of the joint)
            if USE_ROM:
                features[f'{col}_rom'] = df[col].max() - df[col].min()

            # 2. VELOCITY (Speed of movement - critical for Parkinson's)
            if USE_VELOCITY:
                velocity = df[col].diff().abs()
                features[f'{col}_vel_mean'] = velocity.mean()
                features[f'{col}_vel_std'] = velocity.std() # Captures tremors/stuttering

        # 3. SYMMETRY (Left vs Right comparison - critical for Stroke)
        if USE_SYMMETRY:
            # Compare Left and Right Wrist Y-axis Range of Motion
            if 'L_Wrist_y' in df.columns and 'R_Wrist_y' in df.columns:
                l_rom = df['L_Wrist_y'].max() - df['L_Wrist_y'].min()
                r_rom = df['R_Wrist_y'].max() - df['R_Wrist_y'].min()
                features['Wrist_Y_Symmetry'] = abs(l_rom - r_rom)

            # Compare Left and Right Knee Y-axis Range of Motion
            if 'L_Knee_y' in df.columns and 'R_Knee_y' in df.columns:
                l_k_rom = df['L_Knee_y'].max() - df['L_Knee_y'].min()
                r_k_rom = df['R_Knee_y'].max() - df['R_Knee_y'].min()
                features['Knee_Y_Symmetry'] = abs(l_k_rom - r_k_rom)

        return features
    except Exception as e:
        print(f"Error reading {csv_path}: {e}")
        return None

def main():
    csv_files = glob.glob(str(INPUT_CSV_DIR / "*.csv"))
    if not csv_files:
        print(f"Error: No CSV files found in {INPUT_CSV_DIR}")
        return

    X_task1, y_task1 = [], []
    X_task2, y_task2 = [], []

    for csv_file in csv_files:
        basename = Path(csv_file).name
        group_label = basename.split('_')[0]

        summary = create_summary_features(csv_file)
        if not summary: continue

        if group_label in ['E', 'NE']:
            X_task1.append(summary)
            y_task1.append(group_label)
        elif group_label in ['S', 'B', 'P']:
            X_task2.append(summary)
            y_task2.append(group_label)

    cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
    scoring_metrics = ['accuracy', 'f1_macro']
    pipeline = make_pipeline(StandardScaler(), RandomForestClassifier(n_estimators=100, random_state=42))

    print("=" * 60)
    print(f"ACTIVE FEATURES: Baseline={USE_BASELINE}, ROM={USE_ROM}, Velocity={USE_VELOCITY}, Symmetry={USE_SYMMETRY}")
    print("=" * 60)

    # TASK 1
    if len(X_task1) > 0:
        print("\nTASK 1: Expert (E) vs Non-Expert (NE)")
        df_X1 = pd.DataFrame(X_task1)
        scores1 = cross_validate(pipeline, df_X1, LabelEncoder().fit_transform(y_task1), cv=cv, scoring=scoring_metrics, n_jobs=-1)
        print(f"Accuracy:  {scores1['test_accuracy'].mean():.4f} | F-Measure: {scores1['test_f1_macro'].mean():.4f}")

    # TASK 2
    if len(X_task2) > 0:
        print("\nTASK 2: Clinical Classification (Stroke, BackPain, Parkinson)")
        df_X2 = pd.DataFrame(X_task2)
        scores2 = cross_validate(pipeline, df_X2, LabelEncoder().fit_transform(y_task2), cv=cv, scoring=scoring_metrics, n_jobs=-1)
        print(f"Accuracy:  {scores2['test_accuracy'].mean():.4f} | F-Measure: {scores2['test_f1_macro'].mean():.4f}")

if __name__ == "__main__":
    main()
```

```python
%%writefile scripts/feature_extractor.py
import cv2
import argparse
import math
import sys
from pathlib import Path
from mediapipe.tasks.python.vision import pose_landmarker
from mediapipe.tasks.python.vision.core import image as mp_image_lib
from mediapipe.tasks.python.core import base_options
from mediapipe.tasks.python.vision.core import vision_task_running_mode

# Updated for 12 Reference Points (Professor's Task 1)
TRACKED_LANDMARKS = {
    11: "L_Shoulder", 12: "R_Shoulder",
    13: "L_Elbow",    14: "R_Elbow",
    15: "L_Wrist",    16: "R_Wrist",
    23: "L_Hip",      24: "R_Hip",
    25: "L_Knee",     26: "R_Knee",
    27: "L_Ankle",    28: "R_Ankle"
}
OUTPUTS_DIR = Path("outputs")

SCRIPT_DIR = Path(__file__).resolve().parent
DEFAULT_POSE_MODEL_PATH = SCRIPT_DIR / "pose_landmarker_heavy.task"

def _ensure_pose_model() -> Path:
    path = DEFAULT_POSE_MODEL_PATH
    if path.is_file(): return path
    print("Heavy Pose modeli indiriliyor...", flush=True)
    import urllib.request
    urllib.request.urlretrieve("https://storage.googleapis.com/mediapipe-models/pose_landmarker/pose_landmarker_heavy/float16/latest/pose_landmarker_heavy.task", path)
    return path

def video_path_to_csv_basename(video_path: Path, data_path: Path) -> str:
    try: rel = video_path.resolve().relative_to(data_path.resolve())
    except ValueError: rel = video_path
    if len(rel.parts) <= 1: return rel.stem + ".csv"
    return "_".join(rel.parts[:-2] + (rel.stem,)) + ".csv"

def process_one_video(video_path, output_csv_path=None, is_batch=False):
    cap = cv2.VideoCapture(str(video_path))
    if not cap.isOpened(): return 0

    model_bytes = open(_ensure_pose_model(), "rb").read()
    opts = pose_landmarker.PoseLandmarkerOptions(
        base_options=base_options.BaseOptions(model_asset_buffer=model_bytes),
        running_mode=vision_task_running_mode.VisionTaskRunningMode.VIDEO,
        min_pose_detection_confidence=0.7,
        min_tracking_confidence=0.7,
    )
    landmarker = pose_landmarker.PoseLandmarker.create_from_options(opts)

    csv_file = None
    if output_csv_path is not None:
        csv_file = open(output_csv_path, "w", encoding="utf-8")
        header_cols = ["timestamp_sec"]
        for lm_id, name in TRACKED_LANDMARKS.items():
            header_cols.extend([f"{name}_x", f"{name}_y", f"{name}_z"])
        csv_file.write(",".join(header_cols) + "\n")

    frame_idx = 0
    fps = cap.get(cv2.CAP_PROP_FPS)
    if not fps or fps <= 0: fps = 30.0

    try:
        while True:
            ret, frame = cap.read()
            if not ret: break
            frame_idx += 1
            if is_batch and frame_idx % 50 == 0:
                print(f"    -> {frame_idx}. kare isleniyor...", end="\r", flush=True)

            timestamp_ms = int((frame_idx * 1000) / fps)
            rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
            mp_img = mp_image_lib.Image(mp_image_lib.ImageFormat.SRGB, rgb)
            result = landmarker.detect_for_video(mp_img, timestamp_ms)

            row_values = [f"{timestamp_ms / 1000.0:.3f}"]

            if result.pose_world_landmarks and len(result.pose_world_landmarks) > 0:
                world = result.pose_world_landmarks[0]
                for lm_id in TRACKED_LANDMARKS.keys():
                    row_values.extend([f"{world[lm_id].x:.6f}", f"{world[lm_id].y:.6f}", f"{world[lm_id].z:.6f}"])
            else:
                row_values.extend([""] * (len(TRACKED_LANDMARKS) * 3))

            if csv_file: csv_file.write(",".join(row_values) + "\n")

        if is_batch: print(" " * 50, end="\r", flush=True)

    finally:
        cap.release()
        if csv_file: csv_file.close()
        landmarker.close()
    return frame_idx

def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("video", type=str, nargs="?", default=None)
    parser.add_argument("--data-path", type=str, default=None)
    args = parser.parse_args()

    if args.data_path is not None:
        data_path = Path(args.data_path)
        video_files = sorted(data_path.rglob("*.mp4")) + sorted(data_path.rglob("*.avi"))
        OUTPUTS_DIR.mkdir(parents=True, exist_ok=True)
        print(f"\nToplam {len(video_files)} adet video bulundu. GELISMIS VERI CIKARIMI basliyor...\n")

        for i, v in enumerate(video_files, 1):
            out_csv = OUTPUTS_DIR / video_path_to_csv_basename(v, data_path)

            # --- THE FIX: Skip if it already exists ---
            if out_csv.exists() and out_csv.stat().st_size > 0:
                print(f"[{i}/{len(video_files)}] ATLANDI (Zaten hazir): {out_csv.stem}")
                continue
            # ------------------------------------------

            print(f"[{i}/{len(video_files)}] Isleniyor: {out_csv.stem}")
            n = process_one_video(v, out_csv, is_batch=True)
            print(f"  => TAMAMLANDI! ({n} kare)\n")
        print("Tebrikler! Yeni genis CSV veri seti hazirlandi.")

if __name__ == "__main__":
    main()
```

```python
%%writefile scripts/feature_selection_table.py
import pandas as pd
import numpy as np
import glob
import re
import warnings
from pathlib import Path

from sklearn.model_selection import cross_validate, StratifiedKFold
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.pipeline import make_pipeline
from sklearn.feature_selection import SelectKBest, chi2, mutual_info_classif, f_classif
from sklearn.inspection import permutation_importance

from sklearn.linear_model import LogisticRegression, RidgeClassifier, SGDClassifier
from sklearn.svm import SVC
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neural_network import MLPClassifier
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis, QuadraticDiscriminantAnalysis
from sklearn.ensemble import (RandomForestClassifier, GradientBoostingClassifier,
                               AdaBoostClassifier, ExtraTreesClassifier, BaggingClassifier)

warnings.filterwarnings("ignore")

INPUT_CSV_DIR = Path("outputs")
OUTPUT_CSV    = Path("scripts/classifier_feature_selection_results.csv")
K_FEATURES    = 30  # kaç feature seçileceği

# ── Feature extraction ──────────────────────────────────────────────────────
def create_summary_features(csv_path):
    try:
        df = pd.read_csv(csv_path)
        if 'timestamp_sec' in df.columns:
            df = df.drop(columns=['timestamp_sec'])
        df = df.dropna()
        if df.empty or len(df) < 10:
            return None
        features = {}
        for col in df.columns:
            features[f'{col}_max']    = df[col].max()
            features[f'{col}_min']    = df[col].min()
            features[f'{col}_mean']   = df[col].mean()
            features[f'{col}_std']    = df[col].std()
            features[f'{col}_median'] = df[col].median()
            features[f'{col}_rom']    = df[col].max() - df[col].min()
        features['total_frames'] = len(df)
        return features
    except Exception as e:
        print(f"Hata: {csv_path}: {e}")
        return None

# ── Feature selection yöntemleri ────────────────────────────────────────────
def select_features(X, y, method, k):
    """Her yöntem için seçilmiş feature indexlerini döndürür."""
    X_arr = X.values

    if method == "Information Gain":
        scores = mutual_info_classif(X_arr, y, random_state=42)
    elif method == "Gini Decrease":
        rf = RandomForestClassifier(n_estimators=100, random_state=42)
        rf.fit(X_arr, y)
        scores = rf.feature_importances_
    elif method == "ReliefF":
        # ReliefF yaklaşımı: f_classif (ANOVA F-score) ile approximate
        scores, _ = f_classif(X_arr, y)
        scores = np.nan_to_num(scores)
    elif method == "Chi-Square":
        # Chi2 negatif değer kabul etmez → min-max normalize
        X_min = X_arr - X_arr.min(axis=0)
        scores, _ = chi2(X_min, y)
        scores = np.nan_to_num(scores)
    elif method == "FCBF":
        # FCBF yaklaşımı: mutual info + korelasyon filtresi (basit)
        scores = mutual_info_classif(X_arr, y, random_state=0)
        corr   = np.abs(np.corrcoef(X_arr.T))
        np.fill_diagonal(corr, 0)
        redundancy = corr.mean(axis=1)
        scores = scores - 0.5 * redundancy
        scores = np.clip(scores, 0, None)
    elif method == "Information Gain Ratio":
        mi     = mutual_info_classif(X_arr, y, random_state=42)
        # Bölücü: H(X_i) ≈ std tabanlı entropi tahmini
        split  = np.array([max(1e-9, np.std(X_arr[:, i])) for i in range(X_arr.shape[1])])
        scores = mi / split
    else:
        raise ValueError(f"Bilinmeyen yöntem: {method}")

    top_k = np.argsort(scores)[::-1][:k]
    return top_k

# ── Modeller ────────────────────────────────────────────────────────────────
MODELS = {
    "Random Forest":               RandomForestClassifier(n_estimators=100, random_state=42),
    "Extra Trees":                 ExtraTreesClassifier(n_estimators=100, random_state=42),
    "Multi-Layer Perceptron":      MLPClassifier(max_iter=1000, random_state=42),
    "Gradient Boosting":           GradientBoostingClassifier(random_state=42),
    "Logistic Regression":         LogisticRegression(max_iter=1000, random_state=42),
    "Bagging Classifier":          BaggingClassifier(random_state=42),
    "SGD Classifier":              SGDClassifier(random_state=42),
    "Ridge Classifier":            RidgeClassifier(),
    "Linear Discriminant Analysis":LinearDiscriminantAnalysis(),
    "Support Vector Machine":      SVC(random_state=42),
    "AdaBoost":                    AdaBoostClassifier(random_state=42),
    "K-Nearest Neighbors":         KNeighborsClassifier(),
    "Decision Tree":               DecisionTreeClassifier(random_state=42),
    "Gaussian Naive Bayes":        GaussianNB(),
    "Quadratic Discriminant Analysis": QuadraticDiscriminantAnalysis(),
}

FEATURE_METHODS = [
    "Gini Decrease",
    "Information Gain",
    "ReliefF",
    "FCBF",
    "Chi-Square",
    "Information Gain Ratio",
]

def main():
    # ── Veri yükle ──────────────────────────────────────────────────────────
    csv_files = glob.glob(str(INPUT_CSV_DIR / "*.csv"))
    if not csv_files:
        print(f"Hata: {INPUT_CSV_DIR} içinde CSV bulunamadı.")
        return

    print(f"{len(csv_files)} CSV dosyası işleniyor...\n")
    label_regex = re.compile(r"(Es\d+)")
    X_list, y_list = [], []

    for csv_file in csv_files:
        match = label_regex.search(Path(csv_file).name)
        if not match:
            continue
        summary = create_summary_features(csv_file)
        if summary:
            X_list.append(summary)
            y_list.append(match.group(1))

    X  = pd.DataFrame(X_list)
    le = LabelEncoder()
    y  = le.fit_transform(y_list)
    print(f"Dataset: {X.shape[0]} video, {X.shape[1]} feature, {len(set(y_list))} sınıf\n")

    cv = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

    # ── Sonuç tablosu ────────────────────────────────────────────────────────
    results = []

    header = f"{'Classifier':<35}"
    for m in FEATURE_METHODS:
        header += f" | {m[:14]:<14}"
    header += " | Mean Acc"
    print(header)
    print("-" * (35 + 18 * len(FEATURE_METHODS) + 12))

    for name, model in MODELS.items():
        row = {"Classifier": name}
        method_accs = []

        for method in FEATURE_METHODS:
            idx          = select_features(X, y, method, k=K_FEATURES)
            X_sel        = X.iloc[:, idx]
            pipeline     = make_pipeline(StandardScaler(), model)
            scores       = cross_validate(pipeline, X_sel, y, cv=cv,
                                          scoring='accuracy', n_jobs=-1)
            acc          = scores['test_score'].mean()
            row[method]  = round(acc * 100, 2)
            method_accs.append(acc)

        row["Mean Classifier Accuracy"] = round(np.mean(method_accs) * 100, 2)
        results.append(row)

        line = f"{name:<35}"
        for m in FEATURE_METHODS:
            line += f" | {row[m]:>6.2f}%       "
        line += f" | {row['Mean Classifier Accuracy']:>6.2f}%"
        print(line)

    # ── Accuracy'ye göre sırala (yüksekten düşüğe) ──────────────────────────
    df_results = pd.DataFrame(results)
    df_results = df_results.sort_values("Mean Classifier Accuracy", ascending=False)
    df_results = df_results.reset_index(drop=True)

    # ── CSV olarak kaydet ────────────────────────────────────────────────────
    OUTPUT_CSV.parent.mkdir(parents=True, exist_ok=True)
    df_results.to_csv(OUTPUT_CSV, index=False)
    print(f"\n✅ Sonuçlar kaydedildi: {OUTPUT_CSV}")
    print(df_results.to_string(index=False))

if __name__ == "__main__":
    main()
```
